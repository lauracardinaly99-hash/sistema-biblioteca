## **Cadastro de livros(Etapa 2):**

```python
# Lista global que vai armazenar todos os livros
livros = []

def cadastrar_livro():
    """Função responsável por cadastrar um novo livro no sistema"""
    print("\n--- Cadastro de Livro ---")

    # Pedindo os dados para o usuário
    titulo = input("Digite o título do livro: ")
    autor = input("Digite o nome do autor: ")
    ano = input("Digite o ano de publicação: ")

    # Status padrão: livro chega disponível
    disponivel = "Sim"

    # Criando o dicionário que representa o livro
    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": disponivel
    }

    # Adicionando o livro à lista
    livros.append(novo_livro)
    print(f"✅ Livro '{titulo}' cadastrado com sucesso!")

def listar_livros():
    """Função para mostrar todos os livros cadastrados"""
    print("\n--- Lista de Livros Cadastrados ---")

    if len(livros) == 0:
        print("Nenhum livro cadastrado ainda.")
        return

    # Percorrendo a lista e mostrando cada livro
    for i, livro in enumerate(livros, 1):
        print(f"\nLivro {i}:")
        print(f"  Título: {livro['título']}")
        print(f"  Autor: {livro['autor']}")
        print(f"  Ano: {livro['ano']}")
        print(f"  Disponível: {livro['disponível']}")

# ========== PROGRAMA PRINCIPAL ==========
print("Sistema de Biblioteca - Etapa 2")
print("Vamos cadastrar pelo menos 5 livros:\n")

# Cadastrando 5 livros
for _ in range(5):
    cadastrar_livro()

# Mostrando todos os livros cadastrados
listar_livros()
```
