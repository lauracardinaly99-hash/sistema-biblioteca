## **Cadastro de livros(Etapa 2):**

```python
# Lista global que vai armazenar todos os livros
livros = []

def cadastrar_livro():
    """Função responsável por cadastrar um novo livro no sistema"""
    print("\n--- Cadastro de Livro ---")

    # Pedindo os dados para o usuário
    titulo = input("Digite o título do livro: ")
    autor = input("Digite o nome do autor: ")
    ano = input("Digite o ano de publicação: ")

    # Status padrão: livro chega disponível
    disponivel = "Sim"

    # Criando o dicionário que representa o livro
    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": disponivel
    }

    # Adicionando o livro à lista
    livros.append(novo_livro)
    print(f"✅ Livro '{titulo}' cadastrado com sucesso!")

def listar_livros():
    """Função para mostrar todos os livros cadastrados"""
    print("\n--- Lista de Livros Cadastrados ---")

    if len(livros) == 0:
        print("Nenhum livro cadastrado ainda.")
        return

    # Percorrendo a lista e mostrando cada livro
    for i, livro in enumerate(livros, 1):
        print(f"\nLivro {i}:")
        print(f"  Título: {livro['título']}")
        print(f"  Autor: {livro['autor']}")
        print(f"  Ano: {livro['ano']}")
        print(f"  Disponível: {livro['disponível']}")

# ========== PROGRAMA PRINCIPAL ==========
print("Sistema de Biblioteca - Etapa 2")
print("Vamos cadastrar pelo menos 5 livros:\n")

# Cadastrando 5 livros
for _ in range(5):
    cadastrar_livro()

# Mostrando todos os livros cadastrados
listar_livros()
```

### **Consulta e Busca(Etapa 3):**

```python
# Lista global que armazena os livros
livros = []

def cadastrar_livro():
    """Função para cadastrar um novo livro"""
    titulo = input("Título: ")
    autor = input("Autor: ")
    ano = input("Ano: ")

    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": "Sim"
    }
    livros.append(novo_livro)
    print(f"✅ Livro '{titulo}' cadastrado!\n")

def listar_livros():
    """Etapa 3: Listagem geral de todos os livros"""
    print("\n=== Todos os Livros Cadastrados ===")
    if not livros:
        print("Nenhum livro cadastrado ainda.")
        return
    for i, livro in enumerate(livros, 1):
        print(f"{i}. {livro['título']} — {livro['autor']} ({livro['ano']}) | Disponível: {livro['disponível']}")

def buscar_por_titulo():
    """Etapa 3: Buscar livro pelo título"""
    termo = input("\nDigite o título ou parte dele: ").strip().lower()
    encontrados = []
    for livro in livros:
        if termo in livro["título"].lower():
            encontrados.append(livro)
    if encontrados:
        print(f"\n🔍 Encontrado(s) {len(encontrados)} livro(s):")
        for l in encontrados:
            print(f"- {l['título']} — {l['autor']} ({l['ano']})")
    else:
        print("❌ Nenhum livro encontrado com esse título.")

def buscar_por_autor():
    """Etapa 3: Buscar livros pelo autor"""
    termo = input("\nDigite o nome do autor ou parte dele: ").strip().lower()
    encontrados = []
    for livro in livros:
        if termo in livro["autor"].lower():
            encontrados.append(livro)
    if encontrados:
        print(f"\n🔍 Encontrado(s) {len(encontrados)} livro(s):")
        for l in encontrados:
            print(f"- {l['título']} — {l['autor']} ({l['ano']})")
    else:
        print("❌ Nenhum livro encontrado com esse autor.")

# ========== MENU PRINCIPAL ==========
while True:
    print("\n===== SISTEMA DE BIBLIOTECA =====")
    print("1. Cadastrar Livro")
    print("2. Listar Todos os Livros")
    print("3. Buscar por Título")
    print("4. Buscar por Autor")
    print("5. Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        cadastrar_livro()
    elif opcao == "2":
        listar_livros()
    elif opcao == "3":
        buscar_por_titulo()
    elif opcao == "4":
        buscar_por_autor()
    elif opcao == "5":
        print("Saindo do sistema... Até logo!")
        break
    else:
        print("Opção inválida! Tente novamente.")
```

### **Empréstimos e Devoluções(Etapa 4):**

```python
# Lista global que armazena os livros
livros = []

def cadastrar_livro():
    titulo = input("Título: ")
    autor = input("Autor: ")
    ano = input("Ano: ")
    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": "Sim"
    }
    livros.append(novo_livro)
    print(f"✅ Livro '{titulo}' cadastrado!\n")

def listar_livros():
    print("\n--- Acervo ---")
    if not livros:
        print("Nenhum livro cadastrado.\n")
        return
    for i, livro in enumerate(livros, 1):
        print(f"{i}. {livro['título']} — {livro['autor']} ({livro['ano']}) | Disponível: {livro['disponível']}")
    print()

# ========== ETAPA 4 — EMPRÉSTIMO E DEVOLUÇÃO ==========
def emprestar_livro():
    """Altera status de disponível de 'Sim' para 'Não'"""
    titulo_busca = input("Digite o título do livro para empréstimo: ")
    for livro in livros:
        if livro["título"].strip().lower() == titulo_busca.strip().lower():
            if livro["disponível"] == "Sim":
                livro["disponível"] = "Não"
                print(f"📚 Empréstimo realizado! O livro '{livro['título']}' agora está indisponível.\n")
            else:
                print(f"⚠️ Este livro já está emprestado.\n")
            return
    print(f"❌ Livro '{titulo_busca}' não encontrado.\n")

def devolver_livro():
    """Reverte status de disponível para 'Sim'"""
    titulo_busca = input("Digite o título do livro para devolução: ")
    for livro in livros:
        if livro["título"].strip().lower() == titulo_busca.strip().lower():
            if livro["disponível"] == "Não":
                livro["disponível"] = "Sim"
                print(f"📖 Devolução realizada! O livro '{livro['título']}' está disponível novamente.\n")
            else:
                print(f"⚠️ Este livro já está disponível.\n")
            return
    print(f"❌ Livro '{titulo_busca}' não encontrado.\n")

# ========== MENU PRINCIPAL ==========
def menu():
    while True:
        print("==== SISTEMA DE BIBLIOTECA ====")
        print("1. Cadastrar livro")
        print("2. Listar todos os livros")
        print("3. Emprestar livro")
        print("4. Devolver livro")
        print("5. Sair")
        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            cadastrar_livro()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            emprestar_livro()
        elif opcao == "4":
            devolver_livro()
        elif opcao == "5":
            print("Saindo do sistema... Até logo!")
            break
        else:
            print("Opção inválida! Tente novamente.\n")

# Iniciar o programa
menu()
```

### **Persistência com JSON(Etapa 5):**

```python
import json

# Nome do arquivo onde os dados serão salvos
ARQUIVO_LIVROS = "livros.json"

# Lista global que vai armazenar todos os livros
livros = []

def carregar_dados():
    """Carrega os livros do arquivo JSON ao iniciar o programa"""
    global livros
    try:
        with open(ARQUIVO_LIVROS, "r", encoding="utf-8") as arquivo:
            livros = json.load(arquivo)
        print(f"✅ Dados carregados! {len(livros)} livros no sistema.")
    except FileNotFoundError:
        print("📂 Arquivo de dados não encontrado. Iniciando com acervo vazio.")
        livros = []

def salvar_dados():
    """Salva a lista de livros no arquivo JSON"""
    with open(ARQUIVO_LIVROS, "w", encoding="utf-8") as arquivo:
        json.dump(livros, arquivo, ensure_ascii=False, indent=4)
    print("💾 Dados salvos com sucesso!")

def cadastrar_livro():
    """Cadastra um novo livro e salva no arquivo"""
    print("\n--- Cadastro de Livro ---")
    titulo = input("Digite o título do livro: ")
    autor = input("Digite o nome do autor: ")
    ano = input("Digite o ano de publicação: ")

    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": "Sim"
    }

    livros.append(novo_livro)
    salvar_dados()  # Salva automaticamente após cadastrar
    print(f"✅ Livro '{titulo}' cadastrado!")

def listar_livros():
    """Mostra todos os livros cadastrados"""
    print("\n--- Lista de Livros ---")
    if not livros:
        print("Nenhum livro cadastrado.")
        return
    for i, livro in enumerate(livros, 1):
        print(f"\nLivro {i}:")
        print(f"  Título: {livro['título']}")
        print(f"  Autor: {livro['autor']}")
        print(f"  Ano: {livro['ano']}")
        print(f"  Disponível: {livro['disponível']}")

def menu():
    """Menu principal do sistema"""
    while True:
        print("\n===== SISTEMA DE BIBLIOTECA =====")
        print("1. Cadastrar Livro")
        print("2. Listar Todos os Livros")
        print("3. Sair")
        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            cadastrar_livro()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            print("Saindo do sistema... Até logo!")
            break
        else:
            print("❌ Opção inválida! Tente novamente.")

# ========== INICIO DO PROGRAMA ==========
carregar_dados()  # Carrega os dados salvos ao abrir
menu()
```

### **Melhorias sugeridas(Etapa 6):**

```python
import json

ARQUIVO_LIVROS = "livros.json"
livros = []

# ========== FUNÇÕES DE PERSISTÊNCIA ==========
def carregar_dados():
    global livros
    try:
        with open(ARQUIVO_LIVROS, "r", encoding="utf-8") as arquivo:
            livros = json.load(arquivo)
        print(f"✅ Dados carregados! {len(livros)} livros no sistema.")
    except FileNotFoundError:
        print("📂 Arquivo de dados não encontrado. Iniciando com acervo vazio.")
        livros = []

def salvar_dados():
    with open(ARQUIVO_LIVROS, "w", encoding="utf-8") as arquivo:
        json.dump(livros, arquivo, ensure_ascii=False, indent=4)

# ========== FUNÇÕES PRINCIPAIS ==========
def cadastrar_livro():
    print("\n--- Cadastro de Livro ---")
    titulo = input("Título: ")
    autor = input("Autor: ")
    ano = input("Ano: ")

    novo_livro = {
        "título": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": "Sim"
    }
    livros.append(novo_livro)
    salvar_dados()
    print(f"✅ Livro '{titulo}' cadastrado!")

def listar_livros():
    print("\n--- Lista de Livros ---")
    if not livros:
        print("Nenhum livro cadastrado.")
        return
    for i, livro in enumerate(livros, 1):
        print(f"\n[{i}] {livro['título']}")
        print(f"    Autor: {livro['autor']}")
        print(f"    Ano: {livro['ano']}")
        print(f"    Disponível: {livro['disponível']}")

# ========== ETAPA 6 - MELHORIAS ==========
def contador_acervo():
    """Melhoria 1: Mostra estatísticas do acervo"""
    total = len(livros)
    disponiveis = sum(1 for l in livros if l["disponível"] == "Sim")
    emprestados = total - disponiveis

    print("\n📊 CONTADOR DE ACERVO")
    print(f"Total de livros:      {total}")
    print(f"Disponíveis:         {disponiveis}")
    print(f"Emprestados:         {emprestados}")

def remover_livro():
    """Melhoria 2: Remove um livro pelo número"""
    print("\n🗑️ REMOVER LIVRO")
    if not livros:
        print("Nenhum livro para remover.")
        return

    listar_livros()
    try:
        indice = int(input("\nDigite o NÚMERO do livro que deseja remover: ")) - 1

        if 0 <= indice < len(livros):
            livro_removido = livros.pop(indice)
            salvar_dados()
            print(f"✅ Livro '{livro_removido['título']}' foi removido com sucesso!")
        else:
            print("❌ Número inválido!")
    except ValueError:
        print("❌ Digite apenas números!")

# ========== MENU PRINCIPAL ==========
def menu():
    while True:
        print("\n===== SISTEMA DE BIBLIOTECA =====")
        print("1. Cadastrar Livro")
        print("2. Listar Livros")
        print("3. Contador de Acervo")
        print("4. Remover Livro")
        print("5. Sair")

        opcao = input("\nEscolha uma opção: ")

        if opcao == "1":
            cadastrar_livro()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            contador_acervo()
        elif opcao == "4":
            remover_livro()
        elif opcao == "5":
            print("Saindo... Até logo!")
            break
        else:
            print("❌ Opção inválida!")

# ========== EXECUÇÃO ==========
carregar_dados()
menu()
```
